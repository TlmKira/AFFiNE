Python 在浏览器本地运行，适合做数据分析、画图和论文实验记录。

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
df = pd.DataFrame({"x": x, "sin(x)": np.sin(x)})
df.plot(x="x", y="sin(x)")
plt.title("AFFiNE JupyterLite 示例")
plt.show()
```
